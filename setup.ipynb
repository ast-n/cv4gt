## 1 - Download dataset from Roboflow

```python
import getpass
from roboflow import Roboflow

api_key = getpass.getpass("Enter your Roboflow API key: \n")
rf = Roboflow(api_key)
project = rf.workspace("cos40005capstone").project("cv4gt-data")
version = project.version(1)
dataset = version.download("yolov8")
```

## 2 - Download Pre-Trained Models from HuggingFace

```python
import os
from huggingface_hub import hf_hub_download

# Create models directory if it doesn't exist
os.makedirs("models", exist_ok=True)

# Download the models
repo_id = "ast-n/cv4gt"
models = [
    "YOLOv8-cv4gt-bad-dataset_epochs500.pt",
    "YOLOv8-cv4gt-data-11-04_10e"
]

for model_name in models:
    try:
        model_path = hf_hub_download(repo_id=repo_id, filename=model_name, local_dir="models")
        print(f"Downloaded {model_name} to {model_path}")
    except Exception as e:
        print(f"Error downloading {model_name}: {e}")
```
